**Five limitations of Excel for predictive AI in offshore engineering**

Excel is useful for basic calculations and data analysis, but it has limitations when working with the large, complex datasets and computational requirements involved in predictive AI and offshore engineering.

**1. Limited scalability**

Excel becomes difficult to manage as datasets become very large.

Example: A SCADA system collecting data every second from multiple offshore sensors can generate millions of records over time. Managing and analysing this volume in Excel becomes inefficient.

2. Limited automation

Many Excel workflows depend on manually entering data, updating formulas, and refreshing calculations.

Example: An engineer monitoring subsea equipment may need to repeatedly update operating data and calculate equipment conditions. Python can automate this process.

3. Limited machine learning capabilities

Excel provides formulas and analytical tools, but it is not designed as a comprehensive machine-learning development environment.

Example: Predicting equipment failure from historical pressure, temperature, vibration, and flow data requires machine-learning algorithms that can be implemented much more effectively with Python.

4. Reproducibility and version-control challenges

Complex Excel workbooks can contain many interconnected formulas and manual changes, making it difficult to track exactly who changed what and when.

Example: If an engineer changes a pressure calculation formula in a shared offshore engineering workbook, it can be difficult to identify the previous version or reproduce the original calculation.

5. Difficulty handling complex engineering workflows

Large engineering analyses may require data cleaning, statistical analysis, machine learning, visualisation, and automated reporting.

Example: A predictive-maintenance workflow may require importing sensor data, cleaning missing values, calculating engineering features, training a model, and generating predictions. Python can combine these steps into one reproducible workflow.

**Write a Python function that converts temperature from Fahrenheit to Celsius. Include a docstring, type hints, and input validation. Test it with three values including one invalid input.**

*The conversion formula is:*

°C = (°F - 32) × 5/9

````python
def fahrenheit_to_celsius(fahrenheit: [float | int]) -> float:
    """Converts temperature from Fahrenheit to Celsius.

    Args:
        fahrenheit (float | int): Temperature in Fahrenheit.

    Returns:
        float: Temperature in Celsius.

    Raises:
        TypeError: If the input is not a number.
    """
    if not isinstance(fahrenheit, (int, float)):
        raise TypeError("Input must be a number.")
    celsius = (fahrenheit - 32) * 5/9
    return celsius

# Test 1
print("32°F =", fahrenheit_to_celsius(32), "°C")

# Test 2
print("212°F =", fahrenheit_to_celsius(212), "°C")

# Test 3 - invalid input
try:
    print(fahrenheit_to_celsius("100"))
except TypeError as error:
    print("Invalid input:", error)
````

## Question 5

**What is the output of `barlow_wall_thickness(4467, 10.75, 65000)`? Show the manual calculation step by step to verify your answer.**

Assuming the function uses the Barlow wall-thickness formula:

t = (P × D) / (2 × S)

where:

- t = wall thickness
- P = internal pressure
- D = outside diameter
- S = allowable stress

Given:

P = 4467

D = 10.75

S = 65000

Substituting the values:

t = (4467 × 10.75) / (2 × 65000)

First calculate the numerator:

4467 × 10.75 = 48020.25

Then calculate the denominator:

2 × 65000 = 130000

Therefore:

t = 48020.25 / 130000

t = 0.3693865

Therefore, the output is approximately:

**0.3694**

The exact units depend on the units used for pressure, diameter, and allowable stress in the original function.

````python
pressure = 4467
diameter = 10.75
allowable_stress = 65000

wall_thickness = (
    pressure * diameter
    / (2 * allowable_stress)
)

print(f"Barlow wall thickness = {wall_thickness:.4f}")
````

## Question 6

**Write a Python class called `SubseaPipeline` with properties for `length_km`, `od_inches`, `wt_mm`, `material_grade`, and a method that calculates the pipe volume in cubic metres.**

The pipe is treated as a cylinder.

The internal diameter is calculated by subtracting twice the wall thickness from the outside diameter.

````python
import math


class SubseaPipeline:
    """Represent a subsea pipeline and calculate its internal volume."""

    def __init__(
        self,
        length_km: float,
        od_inches: float,
        wt_mm: float,
        material_grade: str
    ):
        self.length_km = length_km
        self.od_inches = od_inches
        self.wt_mm = wt_mm
        self.material_grade = material_grade

    def calculate_volume(self) -> float:
        """Calculate the internal pipe volume in cubic metres."""

        # Convert pipeline length from kilometres to metres
        length_m = self.length_km * 1000

        # Convert outside diameter from inches to millimetres
        od_mm = self.od_inches * 25.4

        # Calculate internal diameter
        id_mm = od_mm - (2 * self.wt_mm)

        # Convert internal diameter from millimetres to metres
        id_m = id_mm / 1000

        # Calculate internal radius
        radius_m = id_m / 2

        # Calculate cylindrical volume
        volume_m3 = math.pi * radius_m ** 2 * length_m

        return volume_m3
````

````python
pipeline = SubseaPipeline(
    length_km=10,
    od_inches=12,
    wt_mm=15,
    material_grade="X65"
)

volume = pipeline.calculate_volume()

print(f"Pipeline internal volume: {volume:.2f} m³")
````

Version control is essential for engineering software because engineering calculations need to be traceable, reproducible, and protected against accidental changes. Git provides a history of changes and allows engineers to work on different versions of software safely.

An engineer accidentally changes a formula used to calculate pipeline pressure. Git records the change, allowing the team to identify what was changed and restore an earlier version if necessary.

**A SCADA system generates 1 data point per second for 5 sensors on a subsea manifold, 24 hours per day, 7 days per week. How many data points are generated per day? Per year? Could Excel handle this volume? Justify your answer.**

The system generates:

1 data point/second × 5 sensors = 5 data points/second.

There are:

60 × 60 × 24 = 86,400 seconds per day.

Therefore:

5 × 86,400 = 432,000 data points per day.

For a 365-day year:

432,000 × 365 = 157,680,000 data points per year.

Therefore:

- **Per day: 432,000 data points**
- **Per year: 157,680,000 data points**

A standard Excel worksheet has a maximum of 1,048,576 rows. Therefore, 157.68 million individual records cannot fit into a single Excel worksheet.

This volume would require multiple worksheets or files and would become difficult to manage efficiently. A database and tools such as Python are more appropriate for processing and analysing this type of continuous sensor data.

````python
data_points_per_second = 5
seconds_per_day = 24 * 60 * 60
days_per_year = 365

data_points_per_day = data_points_per_second * seconds_per_day
data_points_per_year = data_points_per_day * days_per_year

print("Data points per day:", data_points_per_day)
print("Data points per year:", data_points_per_year)
````

**Explain the try/except pattern. Write a function that safely divides two numbers and returns 'Error: Division by zero' if the denominator is zero.**

The `try/except` pattern allows Python to handle errors without terminating the entire program.

The code that might produce an error is placed inside the `try` block. If a specific error occurs, the corresponding `except` block handles it.

This is useful in engineering software because unexpected input or calculation conditions can be handled safely.

````python
def safe_divide(a: float, b: float):
    """Safely divide a by b."""

    try:
        return a / b
    except ZeroDivisionError:
        return "Error: Division by zero"
````

````python
print(safe_divide(10, 2))
print(safe_divide(10, 0))
````
